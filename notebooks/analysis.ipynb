```python
sea_gdp_df = gdp_df[gdp_df["Country"].isin(["Philippines", "Indonesia","Brunei", "Timor-Leste","Cambodia","Laos","Malaysia","Myanmar","Singapore","Thailand","Vietnam"])].set_index("Country")
sea_gdp_df.columns = sea_gdp_df.columns.astype(int)
sea_pop_df.columns = sea_pop_df.columns.astype(int)
print(sea_gdp_df)
print(sea_pop_df)
```

# ANALYSIS

```python
#Mean Population per Country
sea_pop_df_mean = sea_pop_df.mean(axis=1)
sea_pop_df_mean = sea_pop_df_mean.rename("Average Population")
#Mean GDP per Country
sea_gdp_df_mean = sea_gdp_df.mean(axis=1)
sea_gdp_df_mean = sea_gdp_df_mean.rename("Average GDP")
#GDP per capita
sea_gdp_per_capita = (sea_gdp_df * 1000000) / sea_pop_df
sea_gdp_per_capita_percent_change = pd.DataFrame({
    "2020-2021" : ((sea_gdp_per_capita[2021] - sea_gdp_per_capita[2020]) / sea_gdp_per_capita[2021]) * 100,
    "2021-2022" : ((sea_gdp_per_capita[2022] - sea_gdp_per_capita[2021]) / sea_gdp_per_capita[2021]) * 100,
    "2022-2023" : ((sea_gdp_per_capita[2023] - sea_gdp_per_capita[2022]) / sea_gdp_per_capita[2021]) * 100,
    "2023-2024" : ((sea_gdp_per_capita[2024] - sea_gdp_per_capita[2023]) / sea_gdp_per_capita[2021]) * 100,
    "2024-2025" : ((sea_gdp_per_capita[2025] - sea_gdp_per_capita[2024]) / sea_gdp_per_capita[2021]) * 100

})
print(sea_gdp_per_capita_percent_change)
```

# VISUALIZATION

```python
import matplotlib.pyplot as plt
for Country in sea_gdp_per_capita_percent_change.index:
    plt.plot( sea_gdp_per_capita_percent_change.columns, sea_gdp_per_capita_percent_change.loc[Country], label = Country, marker = 'o')
plt.legend(bbox_to_anchor = (1,1),loc = 'best', title = 'SEA COUNTRIES')
plt.title("SEA GDP GROWTH")
plt.ylabel("% Change")
plt.show()
```
